```python
from keras.datasets import imdb
print("IMDB dataset imported successfully.")
```

```python
num_words = 10000
skip_top = 20
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_words, skip_top=skip_top)
print(f"IMDB dataset loaded with {num_words} most common words, skipping the top {skip_top} words.")
print(f"Training sequences: {len(X_train)}")
print(f"Training labels: {len(y_train)}")
print(f"Test sequences: {len(X_test)}")
print(f"Test labels: {len(y_test)}")
```

```python
from keras.preprocessing.sequence import pad_sequences

maxlen = 256

X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)

print(f"Training sequences padded to length: {maxlen}")
print(f"Shape of X_train after padding: {X_train.shape}")
print(f"Shape of X_test after padding: {X_test.shape}")
```

```python
from keras.models import Sequential
from keras.layers import Embedding, LSTM, Dense

embedding_dim = 128

model = Sequential()
model.add(Embedding(num_words, embedding_dim, input_length=maxlen))
model.add(LSTM(128))
model.add(Dense(1, activation='sigmoid'))

print("RNN model architecture built successfully.")
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
print("Model compiled successfully.")

epochs = 5
batch_size = 64

history = model.fit(X_train, y_train, epochs=epochs, batch_size=batch_size, validation_split=0.2)
print("Model trained successfully.")

loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test Accuracy: {accuracy*100:.2f}%")
```
